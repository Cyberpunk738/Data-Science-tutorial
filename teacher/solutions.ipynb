## E · Update a copy, rename and export


```python
shipping_df = df.copy()
shipping_df.loc[shipping_df["Shipped"] == "No", "Shipped"] = "Pending"
shipping_df = shipping_df.rename(columns={"CustomerName": "Customer"})
shipping_df.to_csv(OUTPUT_DIR / "shipping_practice.csv", index=False)
assert "CustomerName" in df.columns
assert "Pending" not in set(df["Shipped"])
shipping_df.head()
```
