## 1. Python basics
Python is a programming language. Its interpreter runs the instructions we write. `print()` displays a value, quotes mark text, and `#` starts a comment.

```python
print("Hello, data science!")
print(2 + 3)
```

### Variables and data types
`=` assigns a value to a name. `int` means whole number, `float` means a number represented with a decimal point, `str` means text and `bool` means True or False.

```python
customer_name = "Alice Wong"
quantity = 3
price = 45.0
shipped = True

print(customer_name)
print(quantity)
print("quantity")
print(type(quantity))
print(type(price))
print(type(customer_name))
print(type(shipped))
```

### Calculations
`+` adds, `-` subtracts, `*` multiplies and `/` divides. This is an arithmetic example, not a calculation of revenue from the CSV.

```python
print(10 + 2)
print(10 - 2)
print(10 * 2)
print(10 / 2)

first_quantity = 3
second_quantity = 5
print(first_quantity + second_quantity)
```

### Lists
Square brackets create a list. Positions start at zero. `len()` counts items.

```python
countries = ["USA", "Canada", "UAE"]
print(countries[0])
print(countries[1])
print(len(countries))
```

### Comparisons and a simple decision
`==` compares for equality; `!=` means not equal. `>=` and `<=` include equality. A colon starts an `if`/`else` block, and indentation groups its instructions.

```python
quantity = 20
print(quantity > 20)
print(quantity >= 20)
print(quantity != 20)
print(quantity == 20)

quantity = 1
if quantity >= 2:
    print("Multiple units")
else:
    print("One unit or fewer")
```

### A dictionary
A dictionary pairs keys with values. Curly braces hold it; a colon links each key to its value.

```python
order = {"CustomerName": "Alice Wong", "Quantity": 3}
print(order["CustomerName"])
```

## 2. Load the orders table with pandas
A library provides ready-made tools. pandas helps us analyse tables. `pd` is its conventional short alias. `df` is our variable name for the DataFrame. A CSV is a text file containing rows and comma-separated fields.

```python
import pandas as pd

df = pd.read_csv("orders.csv")
df.head()
```

### Inspect the table
`df.columns` gives the column labels and has no parentheses. `df.shape` gives (rows, columns). `head()` is a method we call; it shows the first five rows.

```python
print(df.columns)
print(df.shape)
print(len(df))
```

### Select one column and count its entries
One selected column is a Series. Use the exact label `Country`, including its capital C. Its length counts entries, not distinct countries.

```python
df["Country"]
```

```python
len(df["Country"])
```

### Select several columns
The inner brackets make a list of labels. The outer brackets select those columns.

```python
df[["CustomerName", "Country"]].head()
```

```python
df[["Product", "Category", "Quantity"]].head()
```

### Select rows by position
`.iloc[...]` uses integer positions. Zero is the first row. A slice stops before its upper limit. OrderID is a data column, not automatically a row position.

```python
df.iloc[0]
```

```python
df.iloc[0:3]
```

## 3. Filter rows
Create a True/False condition, then keep the matching rows. This selects records without changing them.

```python
electronics = df["Category"] == "Electronics"
print(electronics.head())
df[electronics]
```

The compact form gives the same result:

```python
df[df["Category"] == "Electronics"]
```

### Quantity comparisons
`> 20` excludes 20; `>= 20` includes it. Predict which customers appear before running.

```python
df[df["Quantity"] > 20]
```

```python
df[df["Quantity"] >= 20]
```

```python
df[df["Quantity"] >= 2]
```

```python
df[df["Quantity"] <= 2]
```

```python
df[df["Country"] != "USA"]
```

### Two conditions: AND
`&` requires both conditions to match. Put each condition in parentheses. Use `&` for these pandas conditions instead of Python’s `and` keyword.

```python
df[(df["Category"] == "Electronics") & (df["Country"] == "USA")]
```

### Two conditions: OR
`|` requires at least one condition to match. Rows matching both still appear only once.

```python
df[(df["Category"] == "Electronics") | (df["Country"] == "USA")]
```

### Filter text
`.str` accesses text operations. Use `startswith` and `endswith`. `na=False` treats missing values as not matching. These matches are case-sensitive.

```python
df[df["CustomerName"].str.startswith("A", na=False)]
```

```python
df[df["CustomerName"].str.endswith("son", na=False)]
```

## 4. Locate and update values
`.loc[rows, columns]` selects using labels or a condition. Anna Ivanova exists in this dataset; use her exact recorded name.

```python
df.loc[df["CustomerName"] == "Anna Ivanova", ["CustomerName", "Product"]]
```

Make a copy to experiment. The second `.loc` slot identifies the column to update. Desk Timer is an illustrative replacement for this exercise.

```python
edited_df = df.copy()
edited_df.loc[edited_df["CustomerName"] == "Anna Ivanova", "Product"] = "Desk Timer"
edited_df.loc[edited_df["CustomerName"] == "Anna Ivanova", ["CustomerName", "Product"]]
```

### Replace matching country values
This changes the USA values in the Country column of the copy. It does not rename the column or change other countries.

```python
edited_df.loc[edited_df["Country"] == "USA", "Country"] = "United States"
edited_df.loc[edited_df["Country"] == "United States", ["CustomerName", "Country"]]
```

### Rename a column
Use `columns=` (plural). The dictionary maps the old label to the new one. Keep the returned DataFrame by assigning it.

```python
edited_df = edited_df.rename(columns={"OrderID": "Order ID"})
print(edited_df.columns)
```

## 5. Count and save
`value_counts()` counts how often each label appears. These are order counts, not counts of unique customers. `nunique()` counts distinct non-missing labels.

```python
df["Country"].value_counts()
```

```python
df["Country"].nunique()
```

```python
edited_df["Country"].value_counts()
```

### Save your edited table
This writes `new_file.csv` into the main folder. `index=False` excludes pandas row labels from the CSV. False has a capital F. The original orders.csv is unchanged.

```python
edited_df.to_csv("new_file.csv", index=False)
print("Saved new_file.csv")
```
