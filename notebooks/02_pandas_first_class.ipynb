## 1. Import pandas and load the CSV
`pd` is an alias we choose for pandas. `df` is our variable name for the table. Neither name is a Python keyword. A CSV is a text file containing rows and comma-separated fields.

Run the file-path preparation cell, then the import/read cell.


```python
from pathlib import Path

# Preparation helper: find the CSV whether the kernel starts in the
# repository folder, notebooks folder, or teacher folder.
# We use pathlib from Python's standard library, not another install.
ROOT = next(
    (folder for folder in [Path.cwd(), *Path.cwd().parents]
     if (folder / "data" / "orders.csv").is_file()),
    None,
)
if ROOT is None:
    raise FileNotFoundError("Open the Data-Science-tutorial folder in VS Code.")
DATA_FILE = ROOT / "data" / "orders.csv"
OUTPUT_DIR = ROOT / "outputs"
OUTPUT_DIR.mkdir(exist_ok=True)
```

```python
import pandas as pd

df = pd.read_csv(DATA_FILE)
df.head()
```

`pd.read_csv()` reads a CSV into a DataFrame. `df.head()` shows its first five rows. The last expression in a notebook code cell displays automatically; scripts generally use print().


## 2. Inspect before selecting
`df.columns` is an attribute, so it has no call parentheses. `df.shape` gives (number of rows, number of columns). `len(df)` counts rows.


```python
print(df.columns)
print(df.shape)
print(len(df))
```

## 5. Select a row by integer position
`.iloc` selects by integer position and uses square brackets. Zero means the first row. A slice stops before its upper limit.


```python
df.iloc[0]
```

```python
df.iloc[0:3]
```

The OrderID column contains identifiers such as 1001. Those identifiers are not automatically the table's index or row positions.


## 6. Filter orders by category
First build a True/False condition. Then keep rows where the condition is True. The two-step form helps you see what the nested brackets do.


```python
electronics = df["Category"] == "Electronics"
print(electronics.head())
df[electronics]
```

```python
df[df["Category"] == "Electronics"]
```

## 7. Compare quantities
`>` excludes 20; `>=` includes it. These select rows, rather than changing the values.


```python
df[df["Quantity"] > 20]
```

```python
df[df["Quantity"] >= 20]
```

## 8. Count orders by country
`value_counts()` counts how often each label occurs. These are order counts, not a count of unique people. `nunique()` answers how many distinct non-missing country labels occur.


```python
df["Country"].value_counts()
```

```python
print("Distinct country labels:", df["Country"].nunique())
```

## 9. Save a result
This creates a new file in outputs/. It does not change data/orders.csv. `index=False` leaves pandas row labels out of the exported CSV. Python uses capital F in False.


```python
usa_orders = df[df["Country"] == "USA"]
export_path = OUTPUT_DIR / "usa_orders.csv"
usa_orders.to_csv(export_path, index=False)
print("Saved:", export_path)
```
