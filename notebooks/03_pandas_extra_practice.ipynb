```python
from pathlib import Path

# Preparation helper: find the CSV whether the kernel starts in the
# repository folder, notebooks folder, or teacher folder.
# We use pathlib from Python's standard library, not another install.
ROOT = next(
    (folder for folder in [Path.cwd(), *Path.cwd().parents]
     if (folder / "data" / "orders.csv").is_file()),
    None,
)
if ROOT is None:
    raise FileNotFoundError("Open the Data-Science-tutorial folder in VS Code.")
DATA_FILE = ROOT / "data" / "orders.csv"
OUTPUT_DIR = ROOT / "outputs"
OUTPUT_DIR.mkdir(exist_ok=True)
```

```python
import pandas as pd
df = pd.read_csv(DATA_FILE)
```

## 1. AND: both conditions must be True
Use `&` between pandas conditions, and put **each condition in parentheses**. The Python keyword `and` is not the operator to combine these Series.


```python
df[(df["Category"] == "Electronics") & (df["Country"] == "USA")]
```

## 2. OR: at least one condition must be True
Use `|` (the pipe symbol). Rows satisfying both conditions still appear once.


```python
df[(df["Category"] == "Electronics") | (df["Country"] == "USA")]
```

## 3. Other comparisons
`<`, `<=`, `>=` and `!=` also work. Numbers are unquoted; text labels are quoted.


```python
df[df["Quantity"] >= 2]
```

```python
df[df["Quantity"] <= 2]
```

```python
df[df["Country"] != "USA"]
```

## 4. Names starting with A
`.str` gives access to text operations for a Series. The method is `startswith`, not `startwith`. `na=False` treats missing names as not matching. Matching is case-sensitive.


```python
df[df["CustomerName"].str.startswith("A", na=False)]
```

## 5. Names ending with son
The method is `endswith`, not `endwith`. We choose "son" because it matches several names in this actual dataset.


```python
df[df["CustomerName"].str.endswith("son", na=False)]
```

## 6. Locate rows before editing
`.loc[rows, columns]` uses labels or a Boolean condition. `.iloc` uses integer positions. This example locates the customer by a condition. Anna Ivanova is in the supplied dataset; "Avanova" is not an exact recorded name.


```python
df.loc[df["CustomerName"] == "Anna Ivanova", ["CustomerName", "Product"]]
```

## 7. Update one selected customer's product
`.copy()` makes a separate table for the exercise. `.loc` assigns into a chosen row/column in one operation. The new product text is an illustrative edit. Do not use chained assignment.


```python
edited_df = df.copy()
edited_df.loc[edited_df["CustomerName"] == "Anna Ivanova", "Product"] = "Desk Timer"
edited_df.loc[edited_df["CustomerName"] == "Anna Ivanova", ["CustomerName", "Product"]]
```

## 8. Standardise every matching country entry
This changes all rows whose Country is USA. It is a value replacement in one column, not a column rename. Other countries remain unchanged.


```python
edited_df.loc[edited_df["Country"] == "USA", "Country"] = "United States"
edited_df.loc[edited_df["Country"] == "United States", ["CustomerName", "Country"]]
```

## 9. Rename the OrderID column
Use `columns=` (plural). Here we keep the returned DataFrame. The alternate `inplace=True` form changes that table directly and returns None; do not assign that None back to your variable.


```python
edited_df = edited_df.rename(columns={"OrderID": "Order ID"})
print(edited_df.columns)
```

## 10. Count and export the edited table
Counts reflect the current values, so USA is now United States. `index=False` avoids adding an index field to the CSV. The original CSV is still untouched.


```python
edited_df["Country"].value_counts()
```

```python
edited_path = OUTPUT_DIR / "orders_edited.csv"
edited_df.to_csv(edited_path, index=False)
print("Saved:", edited_path)
```
