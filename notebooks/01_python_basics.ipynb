## 5. Lists and positions
Square brackets create a list. Commas separate items. Positions start at zero. `len()` counts items.


```python
countries = ["USA", "Canada", "UAE"]
print(countries[0])
print(countries[1])
print(len(countries))
```

## 6. Comparisons
`==` asks whether values are equal. `!=` means not equal. `>` and `<` mean greater/less than. `>=` and `<=` include equality. Comparisons return True or False. `=` assigns; `==` compares.


```python
quantity = 20
print(quantity > 20)
print(quantity >= 20)
print(quantity != 20)
print(quantity == 20)
```

## 7. A simple if statement
`if` checks a condition. `else` handles the other case. A colon starts each block and indentation groups its instructions.


```python
quantity = 3
if quantity >= 2:
    print("Multiple units")
else:
    print("One unit or fewer")
```
